```python
# Import Libraries
import pandas as pd
import numpy as np

from scipy.stats import ttest_ind
from scipy.stats import f_oneway

import warnings
warnings.filterwarnings("ignore")
```

```python
# Load Dataset
df = pd.read_csv("../data/insurance.csv")
```

## Hypothesis Testing Concepts

### Null Hypothesis (H₀)

There is no significant difference between groups.

### Alternative Hypothesis (H₁)

There is a significant difference between groups.

### Significance Level

α = 0.05

### Decision Rule

If p-value < 0.05:
Reject H₀

If p-value ≥ 0.05:
Fail to reject H₀

## Test 1 – Diabetes vs PremiumPrice

## Business Question
### Do customers with diabetes pay significantly different premiums?

```python
# Categorical Variable: Diabetes (Yes/No)
diabetes_yes = df[df["Diabetes"] == 1]["PremiumPrice"]
diabetes_no = df[df["Diabetes"] == 0]["PremiumPrice"]
```

```python
# Independent Samples T-Test (Welch's T-Test)
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(
    diabetes_yes,
    diabetes_no,
    equal_var=False  # Welch's t-test
)

print("T Statistic:", round(t_stat, 4))
print("P Value:", p_value)

# Results
if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant difference in premiums between customers with and without diabetes.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in premiums between customers with and without diabetes.")
```

### Observation

The p-value is less than 0.05.

Therefore, the null hypothesis is rejected.

There is a statistically significant difference in PremiumPrice between customers with and without diabetes.

## Test 2 – BloodPressureProblems

## Business Question
### Do customers with blood pressure problems pay significantly different premiums?

```python
# Categorical Variable: Blood Pressure (Yes/No)
BP_yes = df[df["BloodPressureProblems"] == 1]["PremiumPrice"]
BP_no = df[df["BloodPressureProblems"] == 0]["PremiumPrice"]
```

```python
# Independent Samples T-Test (Welch's T-Test)
import scipy.stats as stats

t_stat,p_value = stats.ttest_ind(BP_yes, BP_no, equal_var=False)
print("T Statistic:", round(t_stat, 4))
print("P Value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject the null hypothesis: There is a significant difference in premiums between customers with and without blood pressure problems.")
else:
    print("Fail to reject the null hypothesis: There is no significant difference in premiums between customers with and without blood pressure problems.")
```

### Observation

The p-value is less than 0.05.

Therefore, the null hypothesis is rejected.

There is a statistically significant difference in PremiumPrice between customers with and without blood pressure problem.

## Test 5 – NumberOfMajorSurgeries

## Business Questions
### Do customers with Number of Major Surgeries pay significantly different premiums?

```python
# Categorical Variable: Number of Major Surgeries (0, 1, 2, 3)
group0 = df[df["NumberOfMajorSurgeries"] == 0]["PremiumPrice"]
group1 = df[df["NumberOfMajorSurgeries"] == 1]["PremiumPrice"]
group2 = df[df["NumberOfMajorSurgeries"] == 2]["PremiumPrice"]
group3 = df[df["NumberOfMajorSurgeries"] == 3]["PremiumPrice"]
```

```python
# Anova Test
f_stat, p_value = f_oneway(group0, group1, group2, group3)
print("F Statistic:", round(f_stat, 4))     
print("P Value:", round(p_value, 4))    

if p_value < 0.05:
    print("Reject the null hypothesis: The ANOVA test indicates that at least one surgery group has a significantly different average PremiumPrice.")
else:
    print("Fail to reject the null hypothesis: The ANOVA test indicates no statistically significant difference among surgery groups.")
```

### Observation

The p-value is less than 0.05.

Therefore, the null hypothesis is rejected.

The ANOVA test indicates that at least one surgery group has a significantly different average PremiumPrice.
