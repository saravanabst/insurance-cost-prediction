Insurance companies need to estimate the expected medical cost of new customers.

- Loss of customers if premiums are too high.

The goal of this project is to build a machine learning model that predicts medical insurance charges based on customer information.

Potential business benefits:

- Faster premium estimation

- Better risk assessment
- Reduced manual effort
